### Epsilon Greedy DQN

Consider the following implementation as a reference for implementing other algorithms.

You can also use it as a baseline for comparing the performance of subsequent algorithms.

````python
class EpsGreedyDQNAgent(BaseDQNAgent):
    """
    Epsilon-greedy DQN agent.
    """

    def __init__(self, epsilon: float = 0.1, eps_decay: float = 0.999, eps_min: float = 0.01, **kwargs):
        super().__init__(**kwargs)
        self.epsilon = epsilon
        self.eps_decay = eps_decay
        self.eps_min = eps_min

    def _decay_eps(self):
        """
        Decay the epsilon value.
        """
        self.epsilon = max(self.epsilon * self.eps_decay, self.eps_min)

    def _create_replay_buffer(self, max_size=1000000):
        self.replay_buffer = ReplayBuffer(
            [
                ("state", (self.env.observation_space.shape[0],), torch.float32),
                ("action", (), torch.int64),
                ("reward", (), torch.float32),
                ("next_state", (self.env.observation_space.shape[0],), torch.float32),
                ("done", (), torch.float32),
            ],
            max_size=max_size,
            device=self.device,
        )

    def _create_network(self):
        self.q_network = nn.Sequential(
            nn.Linear(self.env.observation_space.shape[0], 256),
            nn.ReLU(),
            nn.Linear(256, 256),
            nn.ReLU(),
            nn.Linear(256, self.env.action_space.n),
        ).to(self.device)
        self.q_network.apply(
            lambda m: torch.nn.init.orthogonal_(m.weight, gain=torch.nn.init.calculate_gain("relu"))
            if isinstance(m, nn.Linear)
            else None
        )
        self.target_network = nn.Sequential(
            nn.Linear(self.env.observation_space.shape[0], 256),
            nn.ReLU(),
            nn.Linear(256, 256),
            nn.ReLU(),
            nn.Linear(256, self.env.action_space.n),
        ).to(self.device)

    def _compute_loss(self, batch):
        """
        Compute the loss for the DQN agent.
        """
        states = batch["state"]
        actions = batch["action"]
        rewards = batch["reward"]
        next_states = batch["next_state"]
        dones = batch["done"]

        q_values = self.q_network(states).gather(1, actions.unsqueeze(1)).squeeze()
        next_q_values = self.target_network(next_states).max(1)[0]
        expected_q_values = rewards + (1 - dones) * self.gamma * next_q_values

        loss = nn.SmoothL1Loss()(q_values, expected_q_values)
        return loss

    def _act_in_training(self, state):
        """
        Select an action during training.
        """
        self._decay_eps()
        if torch.rand(1).item() < self.epsilon:
            return self.env.action_space.sample()
        else:
            with torch.no_grad():
                q_values = self.q_network(torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device))
                return q_values.argmax().item()

    def _act_in_eval(self, state):
        """
        Select an action during evaluation.
        """
        with torch.no_grad():
            q_values = self.q_network(torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device))
            return q_values.argmax().item()

    def _wandb_train_step_dict(self):
        log_dict = super()._wandb_train_step_dict()
        log_dict["train_step/epsilon"] = self.epsilon
        return log_dict

    def _save_dict(self):
        save_dict = super()._save_dict()
        save_dict["epsilon"] = self.epsilon
        save_dict["eps_decay"] = self.eps_decay
        save_dict["eps_min"] = self.eps_min
        return save_dict
````

### Bootstrap DQN

> Paper: [Deep Exploration via Bootstrapped DQN](https://arxiv.org/abs/1602.04621)

**40 Points**

#### Details

In this algorithm, instead of using a single network, we maintain an ensemble of networks (or a single network with multiple heads). At the start of each training episode, we randomly select one of these networks (heads) and use it to choose actions for the entire episode. This strategy approximates Thompson Sampling for the K-armed Bandit problem, enabling deeper exploration by leveraging the diversity among the ensemble members.

#### Implementation

````python
from torch.distributions import Bernoulli
import torch.nn.functional as F

class MultiHeadQNet(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim, num_heads):
        super().__init__()
        self.num_heads = num_heads
        self.input_dim = input_dim
        self.hidden_dim = hidden_dim
        self.output_dim = output_dim
        
        self.shared_layers = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU()
        )
        
        self.heads = nn.ModuleList([
            nn.Linear(hidden_dim, output_dim) for _ in range(num_heads)
        ])
        
        self._init_weights()
    
    def _init_weights(self):
        for module in self.shared_layers:
            if isinstance(module, nn.Linear):
                torch.nn.init.orthogonal_(module.weight, gain=torch.nn.init.calculate_gain("relu"))
        
        for head in self.heads:
            torch.nn.init.orthogonal_(head.weight, gain=torch.nn.init.calculate_gain("relu"))
    
    def forward(self, x, head_idx=None):
        shared_features = self.shared_layers(x)
        
        if head_idx is not None:
            return self.heads[head_idx](shared_features)
        else:
            outputs = []
            for head in self.heads:
                outputs.append(head(shared_features))
            return torch.stack(outputs, dim=1)

class BootstrapDQNAgent(EpsGreedyDQNAgent):
    """
    Bootstrap DQN agent.
    """

    def __init__(self, k: int = 10, bernoulli_p: float = 0.5, **kwargs):
        self.k = k
        super().__init__(**kwargs)
        self.bernoulli_p = bernoulli_p
        self.bernoulli_dist = Bernoulli(self.bernoulli_p)
        self.current_head = 0

    def _create_network(self):
        self.q_network = MultiHeadQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n,
            num_heads=self.k
        ).to(self.device)
        
        self.target_network = MultiHeadQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n,
            num_heads=self.k
        ).to(self.device)
        
        self.target_network.load_state_dict(self.q_network.state_dict())

    def _create_replay_buffer(self, max_size=1000000):
        self.replay_buffer = ReplayBuffer(
            [
                ("state", (self.env.observation_space.shape[0],), torch.float32),
                ("action", (), torch.int64),
                ("reward", (), torch.float32),
                ("next_state", (self.env.observation_space.shape[0],), torch.float32),
                ("done", (), torch.float32),
                ("mask", (self.k,), torch.float32),
            ],
            max_size=max_size,
            device=self.device,
        )

    def _preprocess_add(self, state, action, reward, next_state, done):
        state = self._state_transformation(state)
        next_state = self._state_transformation(next_state)

        state = torch.tensor(state, dtype=torch.float32, device=self.device)
        action = torch.tensor(action, dtype=torch.long, device=self.device)
        reward = torch.tensor(reward, dtype=torch.float32, device=self.device)
        next_state = torch.tensor(next_state, dtype=torch.float32, device=self.device)
        done = torch.tensor(done, dtype=torch.float32, device=self.device)
        mask = self.bernoulli_dist.sample((self.k,)).float()
        mask = torch.tensor(mask, dtype=torch.float32, device=self.device)
        return {
            "state": state,
            "action": action,
            "reward": reward,
            "next_state": next_state,
            "done": done,
            "mask": mask
        }

    def _compute_loss(self, batch):
        states = batch["state"]
        actions = batch["action"]
        rewards = batch["reward"]
        next_states = batch["next_state"]
        dones = batch["done"]
        masks = batch["mask"]
        
        batch_size = states.shape[0]
        
        q_values_all_heads = self.q_network(states)
        
        with torch.no_grad():
            next_q_values_all_heads = self.target_network(next_states)
            next_q_values_max = next_q_values_all_heads.max(dim=2)[0]
        
        total_loss = 0.0
        
        for head_idx in range(self.k):
            q_values = q_values_all_heads[:, head_idx, :].gather(1, actions.unsqueeze(1)).squeeze()
            
            next_q_values = next_q_values_max[:, head_idx]
            expected_q_values = rewards + (1 - dones) * self.gamma * next_q_values
            
            head_mask = masks[:, head_idx]
            
            if head_mask.sum() > 0:
                masked_q_values = q_values * head_mask
                masked_expected_q_values = expected_q_values * head_mask
                
                head_loss = F.smooth_l1_loss(masked_q_values, masked_expected_q_values, reduction='sum')
                head_loss = head_loss / head_mask.sum()
                total_loss += head_loss
        
        return total_loss / self.k

    def _episode(self):
        self.current_head = torch.randint(0, self.k, (1,)).item()
        
        super()._episode()

    def _act_in_training(self, state):
        self._decay_eps()
        
        if torch.rand(1).item() < self.epsilon:
            return self.env.action_space.sample()
        else:
            with torch.no_grad():
                state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
                q_values = self.q_network(state_tensor, head_idx=self.current_head)
                return q_values.argmax().item()

    def _act_in_eval(self, state):
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            q_values_all_heads = self.q_network(state_tensor)
            q_values_avg = q_values_all_heads.mean(dim=1)
            return q_values_avg.argmax().item()

    def _wandb_train_episode_dict(self):
        log_dict = super()._wandb_train_episode_dict()
        log_dict["train_episode/current_head"] = self.current_head
        return log_dict

    def _save_dict(self):
        save_dict = super()._save_dict()
        save_dict["k"] = self.k
        save_dict["bernoulli_p"] = self.bernoulli_p
        save_dict["current_head"] = self.current_head
        return save_dict
````

### Bootstrap DQN with Randomized Prior Function

> Paper: [Randomized Prior Functions for Deep Reinforcement Learning](https://arxiv.org/abs/1806.03335)

**25 Points**

#### Details

This method is very similar to Bootstrap DQN, but introduces additional **non-trainable** networks (with multiple heads) called random priors. These priors are added to the Q-network outputs to encourage diversity among ensemble members, both across states and over time. During training, the Q-networks learn to compensate for the effect of these fixed random priors, which helps maintain exploration.

##### Notes
- Random prior networks are typically smaller (narrower and shallower) than the main Q-networks, so the Q-networks tend to distill their influence during training.
- There is a $\delta_\mathrm{RPF}$ coefficient to control the strength of the random priors, but for simplicity, you can set $\delta_\mathrm{RPF}=1$ and omit tuning this hyperparameter.

#### Implementation

````python
class PriorMultiHeadQNet(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim, num_heads):
        super().__init__()
        self.num_heads = num_heads
        self.input_dim = input_dim
        self.output_dim = output_dim
        
        prior_hidden_dim = hidden_dim // 2
        
        self.shared_layer = nn.Linear(input_dim, prior_hidden_dim)
        
        self.prior_heads = nn.ModuleList([
            nn.Linear(prior_hidden_dim, output_dim) for _ in range(num_heads)
        ])
        
        self._init_weights()
        self._freeze_parameters()
    
    def _init_weights(self):
        with torch.no_grad():
            torch.nn.init.orthogonal_(self.shared_layer.weight, gain=0.5)
            torch.nn.init.zeros_(self.shared_layer.bias)
            
            for head in self.prior_heads:
                torch.nn.init.orthogonal_(head.weight, gain=0.5)
                torch.nn.init.zeros_(head.bias)
    
    def _freeze_parameters(self):
        for param in self.parameters():
            param.requires_grad = False
    
    def forward(self, x, head_idx=None):
        shared_features = F.relu(self.shared_layer(x))
        
        if head_idx is not None:
            return self.prior_heads[head_idx](shared_features)
        else:
            outputs = []
            for head in self.prior_heads:
                outputs.append(head(shared_features))
            return torch.stack(outputs, dim=1)


class RPFBootstrapDQNAgent(BootstrapDQNAgent):

    def _create_network(self):
        super()._create_network()
        self.prior_network = PriorMultiHeadQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n,
            num_heads=self.k
        ).to(self.device)
        
        self.target_prior_network = PriorMultiHeadQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n,
            num_heads=self.k
        ).to(self.device)
        
        self.target_prior_network.load_state_dict(self.prior_network.state_dict())

    def _act_in_training(self, state):
        self._decay_eps()
        
        if torch.rand(1).item() < self.epsilon:
            return self.env.action_space.sample()
        else:
            with torch.no_grad():
                state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
                
                q_values = self.q_network(state_tensor, head_idx=self.current_head)
                prior_values = self.prior_network(state_tensor, head_idx=self.current_head)
                
                combined_q_values = q_values + 1 * prior_values
                
                return combined_q_values.argmax().item()

    def _act_in_eval(self, state):
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            
            q_values_all_heads = self.q_network(state_tensor)
            prior_values_all_heads = self.prior_network(state_tensor)
            
            combined_q_values = q_values_all_heads + 1 * prior_values_all_heads
            
            q_values_avg = combined_q_values.mean(dim=1)
            return q_values_avg.argmax().item()


    def _compute_loss(self, batch):
        states = batch["state"]
        actions = batch["action"]
        rewards = batch["reward"]
        next_states = batch["next_state"]
        dones = batch["done"]
        masks = batch["mask"]
        
        batch_size = states.shape[0]
        
        q_values_all_heads = self.q_network(states)
        
        prior_values_all_heads = self.prior_network(states)
        
        combined_q_values = q_values_all_heads + 1 * prior_values_all_heads
        
        with torch.no_grad():
            next_q_values_all_heads = self.target_network(next_states)
            next_prior_values_all_heads = self.target_prior_network(next_states)
            
            next_combined_q_values = next_q_values_all_heads + 1 * next_prior_values_all_heads
            
            next_q_values_max = next_combined_q_values.max(dim=2)[0]
        
        total_loss = 0.0
        
        for head_idx in range(self.k):
            q_values = combined_q_values[:, head_idx, :].gather(1, actions.unsqueeze(1)).squeeze()
            
            next_q_values = next_q_values_max[:, head_idx]
            expected_q_values = rewards + (1 - dones) * self.gamma * next_q_values
            
            head_mask = masks[:, head_idx]
            
            if head_mask.sum() > 0:
                masked_q_values = q_values * head_mask
                masked_expected_q_values = expected_q_values * head_mask
                
                head_loss = F.smooth_l1_loss(masked_q_values, masked_expected_q_values, reduction='sum')
                head_loss = head_loss / head_mask.sum()
                total_loss += head_loss
        
        return total_loss / self.k
````

### Uncertainty Estimation for Sample Efficient RPF Bootstrap DQN

> Paper: [Sample Efficient Deep Reinforcement Learning via Uncertainty Estimation](https://arxiv.org/abs/2201.01666)

**35 Points**

#### Details

This method does not explicitly address the exploration problem but focuses on improving sample efficiency.

By maintaining an ensemble of Q-networks, multiple Q-values can be computed for each state-action pair. This enables the estimation of uncertainty in the target values. Using this uncertainty, a weighted loss is calculated, where the weights are inversely proportional to the uncertainty. The more confident we are about a target, the higher its weight during the update.

Logging Effective Batch Size (EBS) helps you to tune $\xi$ parameter:

$$
EBS = \dfrac{\left(\sum^K_k w_k\right)^2}{\sum_k^K w_k^2} = \dfrac{\left(\sum_k^K \dfrac{1}{(\sigma_k^2 + \xi)}\right)^2}{\sum_k^K\dfrac{1}{(\sigma_k^2 + \xi)^2}} 
$$

#### Bonus

1. Use the minimum Effective Batch Size (EBS) as a hyperparameter instead of $\xi$, and numerically calculate $\xi$ during each training step based on the minimum EBS. (5 points)
2. Implement the complete IV-DQN algorithm as described in the appendix of the original paper. (15 points)

#### Implementation

````python
class UEBootstrapDQNAgent(RPFBootstrapDQNAgent):
    """
    Uncertainty Estimation Bootstrap DQN Agent using minimum Effective Batch Size (EBS)
    as hyperparameter instead of xi. Calculates xi dynamically during training.
    """
    def __init__(self, min_effective_batch_size: float = 16.0, xi_init: float = 0.1, **kwargs):
        super().__init__(**kwargs)
        self.min_effective_batch_size = min_effective_batch_size
        self.xi = xi_init
        
        self.last_uncertainty_weights = None
        self.last_effective_batch_size = None
        self.last_mean_uncertainty = None
        self.last_computed_xi = None
        
        self.xi_update_rate = 0.01
        self.xi_min = 1e-6
        self.xi_max = 100.0

    def _compute_dynamic_xi(self, target_variance, batch_size):
        uncertainty_weights = batch_size / (target_variance + self.xi)
        uncertainty_weights = uncertainty_weights * batch_size / uncertainty_weights.sum()
        current_ebs = self._compute_effective_batch_size_from_weights(uncertainty_weights)
        
        if current_ebs < self.min_effective_batch_size:
            xi_low = self.xi_min
            xi_high = self.xi_max
            target_ebs = self.min_effective_batch_size
            
            for _ in range(20):
                xi_mid = (xi_low + xi_high) / 2
                
                test_weights = batch_size / (target_variance + xi_mid)
                test_weights = test_weights * batch_size / test_weights.sum()
                test_ebs = self._compute_effective_batch_size_from_weights(test_weights)
                
                if abs(test_ebs - target_ebs) < 0.1:
                    break
                elif test_ebs < target_ebs:
                    xi_high = xi_mid
                else:
                    xi_low = xi_mid
            
            new_xi = xi_mid
        else:
            new_xi = self.xi
        
        self.xi = self.xi * (1 - self.xi_update_rate) + new_xi * self.xi_update_rate
        self.xi = torch.clamp(torch.tensor(self.xi), self.xi_min, self.xi_max).item()
        
        return self.xi
        
    def _compute_effective_batch_size_from_weights(self, weights):
        w_flat = weights.contiguous().reshape(-1)
        sum_weights = w_flat.sum()
        sum_weights_squared = (w_flat ** 2).sum()
        
        if sum_weights_squared > 0:
            ebs = (sum_weights ** 2) / sum_weights_squared
            return ebs.item()
        else:
            return 0.0
            
    def _compute_loss(self, batch):
        states = batch["state"]
        actions = batch["action"]
        rewards = batch["reward"]
        next_states = batch["next_state"]
        dones = batch["done"]
        head_mask = batch["mask"]
        
        batch_size = states.shape[0]
        
        q_values_all_heads = self.q_network(states)
        prior_values_all_heads = self.prior_network(states)
        combined_q_values = q_values_all_heads + 1 * prior_values_all_heads
        
        actions_expanded = actions.unsqueeze(1).unsqueeze(2).expand(-1, self.k, -1)
        current_q_values = combined_q_values.gather(2, actions_expanded).squeeze(2)
        
        with torch.no_grad():
            next_q_values_all_heads = self.target_network(next_states)
            next_prior_values_all_heads = self.target_prior_network(next_states)
            next_combined_q_values = next_q_values_all_heads + 1 * next_prior_values_all_heads
            
            main_next_q_values = self.q_network(next_states)
            main_next_prior_values = self.prior_network(next_states)
            main_next_combined = main_next_q_values + 1 * main_next_prior_values
            
            next_actions = main_next_combined.mean(dim=1).argmax(dim=1)
            next_actions_expanded = next_actions.unsqueeze(1).unsqueeze(2).expand(-1, self.k, -1)
            next_q_values = next_combined_q_values.gather(2, next_actions_expanded).squeeze(2)
            
            target_q_values = rewards.unsqueeze(-1) + (1 - dones.unsqueeze(-1)) * self.gamma * next_q_values
        
        target_variance = target_q_values.var(dim=1, keepdim=True)
        
        self.last_computed_xi = self._compute_dynamic_xi(target_variance, batch_size)
        
        uncertainty_weights = 1.0 / (target_variance + self.xi)
        uncertainty_weights = uncertainty_weights * batch_size / uncertainty_weights.sum()
        uncertainty_weights = uncertainty_weights.expand(-1, self.k)
        
        losses = []
        total_loss = 0.0
        
        for head_idx in range(self.k):
            head_predictions = current_q_values[:, head_idx]
            head_targets = target_q_values[:, head_idx]
            head_weights = uncertainty_weights[:, head_idx]
            
            head_loss = F.smooth_l1_loss(head_predictions, head_targets, reduction='none')
            weighted_head_loss = (head_loss * head_weights).mean()
            
            losses.append(weighted_head_loss)
            total_loss += weighted_head_loss
        
        self.last_uncertainty_weights = uncertainty_weights.detach()
        self.last_mean_uncertainty = target_variance.mean().item()
        self.last_effective_batch_size = self._compute_effective_batch_size_from_weights(uncertainty_weights)
        
        return total_loss / len(losses) if losses else torch.tensor(0.0, device=self.device)

    def _save_dict(self):
        save_dict = super()._save_dict()
        save_dict.update({
            "min_effective_batch_size": self.min_effective_batch_size,
            "xi": self.xi,
            "xi_update_rate": self.xi_update_rate,
            "xi_min": self.xi_min,
            "xi_max": self.xi_max
        })
        return save_dict

    def _wandb_train_step_dict(self):
        log_dict = super()._wandb_train_step_dict()
        
        log_dict.update({
            "train/min_effective_batch_size_target": self.min_effective_batch_size,
            "train/computed_xi": self.xi,
        })
        
        if self.last_effective_batch_size is not None:
            log_dict["train/effective_batch_size"] = self.last_effective_batch_size
            log_dict["train/ebs_vs_target_ratio"] = self.last_effective_batch_size / self.min_effective_batch_size
            
        if self.last_mean_uncertainty is not None:
            log_dict["train/mean_uncertainty"] = self.last_mean_uncertainty
            
        if self.last_uncertainty_weights is not None:
            weights = self.last_uncertainty_weights
            log_dict.update({
                "train/uncertainty_weights_mean": weights.mean().item(),
                "train/uncertainty_weights_std": weights.std().item(),
                "train/uncertainty_weights_min": weights.min().item(),
                "train/uncertainty_weights_max": weights.max().item()
            })
        
        return log_dict
````

### IV-DQN

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.distributions import Categorical
import numpy as np


class IVQNet(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.input_dim = input_dim
        self.hidden_dim = hidden_dim
        self.output_dim = output_dim
        
        self.shared_layers = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU()
        )
        
        self.q_head = nn.Linear(hidden_dim, output_dim)
        
        self.info_head = nn.Linear(hidden_dim, output_dim)
        
        self._init_weights()
    
    def _init_weights(self):
        for module in self.shared_layers:
            if isinstance(module, nn.Linear):
                torch.nn.init.orthogonal_(module.weight, gain=torch.nn.init.calculate_gain("relu"))
        
        torch.nn.init.orthogonal_(self.q_head.weight, gain=torch.nn.init.calculate_gain("relu"))
        torch.nn.init.orthogonal_(self.info_head.weight, gain=0.1)
    
    def forward(self, x):
        shared_features = self.shared_layers(x)
        q_values = self.q_head(shared_features)
        info_values = self.info_head(shared_features)
        return q_values, info_values


class IVDQNAgent(EpsGreedyDQNAgent):
    
    def __init__(self, 
                 beta: float = 1.0,           # Information bonus coefficient
                 info_lr: float = 1e-4,       # Learning rate for information values
                 info_update_freq: int = 1,   # Frequency of information value updates
                 exploration_type: str = "ucb", # "ucb" or "thompson"
                 **kwargs):
        self.beta = beta
        self.info_lr = info_lr
        self.info_update_freq = info_update_freq
        self.exploration_type = exploration_type
        super().__init__(**kwargs)
        
        self._info_update_step = 0
        
        self.recent_errors = []
        self.max_error_history = 1000
        
    def _create_network(self):
        """Create the IV-DQN network with both Q-value and information value heads."""
        self.q_network = IVQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n
        ).to(self.device)
        
        self.target_network = IVQNet(
            input_dim=self.env.observation_space.shape[0],
            hidden_dim=256,
            output_dim=self.env.action_space.n
        ).to(self.device)
        
        self.target_network.load_state_dict(self.q_network.state_dict())
        self.target_network.eval()
    
    def _create_optimizer(self):
        """Create optimizers for both Q-values and information values."""
        q_params = list(self.q_network.shared_layers.parameters()) + list(self.q_network.q_head.parameters())
        info_params = list(self.q_network.shared_layers.parameters()) + list(self.q_network.info_head.parameters())
        
        self.optimizer = torch.optim.Adam(q_params, lr=self.learning_rate)
        self.info_optimizer = torch.optim.Adam(info_params, lr=self.info_lr)
    
    def _compute_loss(self, batch):
        """
        Compute the combined loss for Q-values and information values.
        """
        states = batch["state"]
        actions = batch["action"]
        rewards = batch["reward"]
        next_states = batch["next_state"]
        dones = batch["done"]
        
        q_values, info_values = self.q_network(states)
        
        q_values_taken = q_values.gather(1, actions.unsqueeze(1)).squeeze()
        
        with torch.no_grad():
            next_q_values, _ = self.target_network(next_states)
            next_q_max = next_q_values.max(1)[0]
            target_q_values = rewards + (1 - dones) * self.gamma * next_q_max
        
        q_loss = F.smooth_l1_loss(q_values_taken, target_q_values)
        
        info_loss = self._compute_info_loss(batch, q_values, info_values, target_q_values)
        
        total_loss = q_loss + info_loss
        
        prediction_errors = torch.abs(q_values_taken - target_q_values)
        self.recent_errors.extend(prediction_errors.detach().cpu().numpy())
        if len(self.recent_errors) > self.max_error_history:
            self.recent_errors = self.recent_errors[-self.max_error_history:]
        
        return total_loss
    
    def _compute_info_loss(self, batch, q_values, info_values, target_q_values):
        """
        Compute the information value loss based on prediction uncertainty.
        """
        states = batch["state"]
        actions = batch["action"]
        
        info_values_taken = info_values.gather(1, actions.unsqueeze(1)).squeeze()
        
        q_values_taken = q_values.gather(1, actions.unsqueeze(1)).squeeze()
        prediction_errors = torch.abs(q_values_taken - target_q_values)
        
        target_info_values = prediction_errors.detach()
        
        info_loss = F.mse_loss(info_values_taken, target_info_values)
        
        return info_loss
    
    def learn(self, batch_size=None):
        """
        Perform learning update with separate optimization for Q-values and information values.
        """
        self._loss = None
        if not self.training:
            print("Agent is not in training mode. It can not learn.")
            return

        if batch_size is None:
            batch_size = self.default_batch_size
        if len(self.replay_buffer) < batch_size or self._training_step < self.start_training_after:
            return

        batch = self.replay_buffer.sample(batch_size)
        if self._norm_rew:
            batch["reward"] = self._normalize_reward(batch["reward"])
        
        loss = self._compute_loss(batch)
        
        self.optimizer.zero_grad()
        loss.backward(retain_graph=True)
        
        if self.gradient_norm_clip is not None:
            torch.nn.utils.clip_grad_norm_(self._learnable_parameters(), self.gradient_norm_clip)

        self.grad_norm = 0
        for param in self._learnable_parameters():
            if param.grad is not None:
                self.grad_norm += param.grad.detach().data.norm(2).item() ** 2.0
        self.grad_norm = self.grad_norm ** (1.0 / 2.0)
        
        self.optimizer.step()
        
        if self._training_step % self.info_update_freq == 0:
            self.info_optimizer.step()
            self._info_update_step += 1
        
        self._soft_update_target_network()

        self._loss = loss.item()
        self._episode_loss += self._loss

        log_dict = self._wandb_train_step_dict()
        if self.wandb_run is not None:
            self.wandb_run.log(log_dict, step=self._training_step)
    
    def _act_in_training(self, state):
        """
        Select action during training using information-guided exploration.
        """
        self._decay_eps()
        
        if torch.rand(1).item() < self.epsilon:
            return self.env.action_space.sample()
        else:
            return self._select_action_with_info(state)
    
    def _act_in_eval(self, state):
        """
        Select action during evaluation (greedy with respect to Q-values).
        """
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            q_values, _ = self.q_network(state_tensor)
            return q_values.argmax().item()
    
    def _select_action_with_info(self, state):
        """
        Select action using information values to guide exploration.
        """
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            q_values, info_values = self.q_network(state_tensor)
            
            if self.exploration_type == "ucb":
                action_values = q_values + self.beta * info_values
                return action_values.argmax().item()
            
            elif self.exploration_type == "thompson":
                uncertainties = torch.clamp(info_values, min=0.01)
                
                sampled_values = torch.normal(q_values, torch.sqrt(uncertainties))
                return sampled_values.argmax().item()
            
            else:
                raise ValueError(f"Unknown exploration type: {self.exploration_type}")
    
    def _wandb_train_step_dict(self):
        """Add IV-DQN specific metrics to logging."""
        log_dict = super()._wandb_train_step_dict()
        
        log_dict["train_step/beta"] = self.beta
        log_dict["train_step/info_update_step"] = self._info_update_step
        
        if hasattr(self, 'recent_errors') and self.recent_errors:
            log_dict["train_step/mean_prediction_error"] = np.mean(self.recent_errors)
            log_dict["train_step/std_prediction_error"] = np.std(self.recent_errors)
        
        return log_dict
    
    def _save_dict(self):
        """Add IV-DQN specific parameters to save dictionary."""
        save_dict = super()._save_dict()
        save_dict["beta"] = self.beta
        save_dict["info_lr"] = self.info_lr
        save_dict["info_update_freq"] = self.info_update_freq
        save_dict["exploration_type"] = self.exploration_type
        save_dict["_info_update_step"] = self._info_update_step
        return save_dict
    
    def get_info_values(self, state):
        """
        Get information values for a given state (useful for analysis).
        """
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            _, info_values = self.q_network(state_tensor)
            return info_values.squeeze().cpu().numpy()
    
    def get_q_and_info_values(self, state):
        """
        Get both Q-values and information values for a given state.
        """
        with torch.no_grad():
            state_tensor = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(self.device)
            q_values, info_values = self.q_network(state_tensor)
            return q_values.squeeze().cpu().numpy(), info_values.squeeze().cpu().numpy()
````
